```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries Imported Successfully")
```

```python
def f(x):
    return x**2 + 2*x + 1

points = [0, 1, 3]

for x in points:
    derivative = (f(x + 0.0001) - f(x)) / 0.0001
    print(f"x = {x}, Numerical Derivative = {derivative:.2f}")
```

```python
x = np.linspace(-4, 4, 400)
y = f(x)

plt.figure(figsize=(8,5))
plt.plot(x, y, label="f(x)=x²+2x+1")

for p in [0, 1, 3]:
    slope = 2*p + 2
    y_point = f(p)

    tangent = slope * (x - p) + y_point

    plt.plot(x, tangent, "--")
    plt.scatter(p, y_point)

plt.title("Function and Tangent Lines")
plt.grid(True)
plt.show()
```

# Task 1 Observations

1. The slope increases as x increases.
2. Positive slope means the function is increasing.
3. Near-zero slope indicates the optimization process is close to a minimum.

```python
import pandas as pd

def loss(theta):
    return (theta - 3.5)**2 + 1

def gradient(theta):
    return 2 * (theta - 3.5)

theta = 10
learning_rate = 0.1

history = []

for i in range(15):
    grad = gradient(theta)
    current_loss = loss(theta)

    history.append([
        i,
        theta,
        current_loss,
        grad
    ])

    theta = theta - learning_rate * grad

df = pd.DataFrame(
    history,
    columns=[
        "Iteration",
        "Theta",
        "Loss",
        "Gradient"
    ]
)

df
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    df["Iteration"],
    df["Loss"],
    marker="o"
)

plt.title("Gradient Descent Loss Curve")
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.grid(True)

plt.show()
```

```python
theta = 10
learning_rate = 0.01

losses_low = []

for i in range(15):
    losses_low.append(loss(theta))
    theta = theta - learning_rate * gradient(theta)
```

```python
theta = 10
learning_rate = 0.5

losses_high = []

for i in range(15):
    losses_high.append(loss(theta))
    theta = theta - learning_rate * gradient(theta)
```

```python
plt.figure(figsize=(8,5))

plt.plot(losses_low, marker="o", label="LR = 0.01")
plt.plot(losses_high, marker="s", label="LR = 0.5")

plt.title("Learning Rate Comparison")
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()
```

# Learning Rate Comparison

1. A low learning rate converges slowly.
2. A higher learning rate reaches the minimum faster.
3. Very large learning rates may overshoot the minimum and become unstable.

```python
uniform_data = np.random.uniform(
    low=0,
    high=100,
    size=1000
)

normal_data = np.random.normal(
    loc=50,
    scale=15,
    size=1000
)

print("Data Generated Successfully")
```

```python
print("Uniform Distribution")
print("Mean:", np.mean(uniform_data))
print("Standard Deviation:", np.std(uniform_data))

print("\nNormal Distribution")
print("Mean:", np.mean(normal_data))
print("Standard Deviation:", np.std(normal_data))
```

```python
plt.figure(figsize=(8,5))

plt.hist(
    uniform_data,
    bins=20
)

plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(
    normal_data,
    bins=20
)

plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

# Interpretation

1. Uniform distribution spreads values evenly across the range.

2. Normal distribution concentrates values around the mean and forms a bell-shaped curve.

3. The mean represents the average value, while standard deviation measures variability.

4. The normal distribution is commonly observed in real-world datasets.

5. Understanding distributions helps in machine learning and statistical analysis.
